## 6. Visualizaciones exploratorias

Genere visualizaciones que permitan comprender mejor los datos.

Cada gráfico debe incluir:

- título;
- interpretación;
- explicación de su relevancia.


```python
#Visualizaciones

# Configuración de estilo
sb.set_theme(style="whitegrid")
fig, axes = plt.subplots(1, 2, figsize=(18, 5))

# 1. Gráfico: Número de clientes por región con leyenda de Género
sb.countplot(
    data=df,
    x='region',
    hue='gender',
    ax=axes[0],
    palette='Blues_r'
)
axes[0].set_title('Número de Clientes por Región y Género')
axes[0].set_xlabel('Región')
axes[0].set_ylabel('Cantidad de Clientes')
axes[0].legend(title='Genero')

# 2. Gráfico: Total de compras por región con leyenda de Nivel Educacional
compras_region_edu = df.groupby(['region', 'education'])['total_spent_clp'].sum().reset_index()

sb.barplot(
    data=compras_region_edu,
    x='region',
    y='total_spent_clp',
    hue='education',
    ax=axes[1],
    palette='Greens_r'
)
axes[1].set_title('Total Gastado por Región y Nivel Educacional (CLP)')
axes[1].set_xlabel('Región')
axes[1].set_ylabel('Total Gastado (CLP)')
axes[1].legend(title='Educación')

plt.tight_layout()
plt.show()
```

```python
# 1. Crear los intervalos de 10 años (ejemplo: de 20 a 80 años)
rangos = [20, 30, 40, 50, 60, 70, 80, float('inf')]
etiquetas = ['20-29', '30-39', '40-49', '50-59', '60-69', '70-79','80+']

# Crear la nueva columna con el rango de edad
df['rango_edad'] = pd.cut(df['age'], bins=rangos, labels=etiquetas, right=False)

# 2. Graficar en barras horizontales
sb.set_theme(style="whitegrid")
plt.figure(figsize=(8, 5))

ax = sb.countplot(
    data=df,
    y='rango_edad',
    hue='rango_edad',
    legend=False,
    palette='Blues_r'
)

plt.title('Cantidad de Clientes por Rango de Edad')
plt.xlabel('Cantidad de Clientes')
plt.ylabel('Rango de Edad')

# Añadir la cantidad exacta al lado de cada barra
for container in ax.containers:
    ax.bar_label(container, padding=3, fontsize=10)

plt.tight_layout()
plt.show()
```

Genere 2 graficos de histograma y un grfico de barras que permiten perfilar a los clientes. "Número de Clientes por Región y Género", permite determinat como de distribuyen los clientes en terminos de genero por region. De la misma manera 'Total Gastado por Región y Nivel Educacional (CLP)', nos da una idea de la relacion entre nivel de educación y el nivel de gasto. Por ultimo 'Cantidad de Clientes por Rango de Edad' identifica el grupo etario donde se concentra el mayor numero de clientes.

La relevancia de estos, es que permiten perfilar a los clientes, asi es posible enfocar campañas en los grupos de clientes mas relevantes a nivel de ventas o generar campañas para los clientes que estan comprando menos.
